````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (10, 4)
pd.set_option('display.max_columns', 20)
````

## <font color='steelblue'>1. El histograma</font>

````python
salarios = pd.read_csv("salarios_redaccion.csv")
encuesta = pd.read_csv("encuesta_medios.csv")
````

Un **histograma** divide el rango de la variable en intervalos contiguos y dibuja una barra con el número de casos de cada uno. A diferencia del gráfico de barras, **las barras van pegadas**, porque el eje es continuo.

````python
plt.hist(salarios["salario_bruto_anual"], bins=20, color="steelblue", edgecolor="white")
plt.xlabel("Salario bruto anual (€)")
plt.ylabel("Número de personas")
plt.title("Salarios en una redacción (datos simulados, n = 420)")
plt.show()
````

**Cómo leerlo**: la mayoría cobra entre 20.000 y 35.000 euros, pero la cola derecha es larga: unos pocos sueldos muy altos (la dirección). Es una distribución **asimétrica a la derecha**, la forma típica de salarios, rentas y audiencias.

## <font color='steelblue'>2. Cuántos intervalos</font>

El número de intervalos (`bins`) cambia mucho la imagen. Pocos ocultan detalles; demasiados dibujan ruido.

````python
fig, axes = plt.subplots(1, 3, figsize=(14, 3.5), sharey=False)
for ax, b in zip(axes, [5, 20, 80]):
    ax.hist(salarios["salario_bruto_anual"], bins=b, color="steelblue", edgecolor="white")
    ax.set_title(f"bins = {b}")
plt.tight_layout()
plt.show()
````

No hay un número correcto: prueba varios y quédate con el que muestra la forma **sin inventar picos**. Una regla de partida es la raíz cuadrada del número de casos (√420 ≈ 20).

## <font color='steelblue'>3. Formas de una distribución</font>

Las formas que más verás:

- **Simétrica**: los valores se reparten igual a ambos lados del centro (la confianza en los medios, aproximadamente).
- **Asimétrica a la derecha**: cola larga hacia los valores altos (salarios, minutos de consumo, compartidos en redes).
- **Bimodal**: dos picos, señal de que se mezclan **dos grupos** distintos.

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 3.5))
axes[0].hist(encuesta["confianza_medios"].dropna(), bins=range(0, 12), color="steelblue", edgecolor="white", align="left")
axes[0].set_title("Confianza en los medios (0-10): casi simétrica")
axes[1].hist(encuesta["minutos_noticias_dia"].dropna(), bins=30, color="darkorange", edgecolor="white")
axes[1].set_title("Minutos al día con noticias: asimétrica a la derecha")
plt.tight_layout()
plt.show()
````

Ahora mira qué pasa si dibujamos juntos los sueldos de la **dirección** y de los **community managers**: aparecen dos montañas separadas. Un histograma bimodal te avisa de que no tiene sentido dar una sola cifra resumen.

````python
dos = salarios[salarios["puesto"].isin(["Community manager", "Dirección", "Redactor/a jefe"])]
plt.hist(dos["salario_bruto_anual"], bins=25, color="grey", edgecolor="white")
plt.title("Mezcla de puestos: más de un pico")
plt.xlabel("Salario bruto anual (€)")
plt.show()
````

## <font color='steelblue'>4. Comparar dos grupos</font>

**Qué vamos a hacer**: comparar los salarios de mujeres y hombres con dos histogramas superpuestos y semitransparentes (`alpha`). `density=True` pone los dos en la misma escala aunque los grupos tengan tamaños distintos.

````python
for sexo, color in [("Mujer", "purple"), ("Hombre", "teal")]:
    datos = salarios.loc[salarios["sexo"] == sexo, "salario_bruto_anual"]
    plt.hist(datos, bins=25, alpha=0.5, color=color, label=f"{sexo} (n = {len(datos)})", density=True)
plt.legend()
plt.xlabel("Salario bruto anual (€)")
plt.title("Distribución de salarios por sexo (datos simulados)")
plt.show()
````

Las dos distribuciones se solapan mucho; a simple vista cuesta ver diferencias. En los cuadernos 8 a 10 aprenderás a resumirlas con números, y en el 9 con un diagrama de caja, que compara grupos mucho mejor.

## <font color='steelblue'>Resumen: ideas clave</font>

- El **histograma** muestra cómo se reparten los valores de una variable cuantitativa; sus barras van pegadas.
- El número de intervalos cambia la imagen: prueba varios; √n es un buen punto de partida.
- Formas típicas: **simétrica**, **asimétrica a la derecha** (salarios, audiencias) y **bimodal** (mezcla de grupos).
- Para comparar grupos de distinto tamaño usa `density=True` y transparencia (`alpha`).
- Mira siempre la distribución **antes** de resumirla con una sola cifra.

## <font color='steelblue'>Glosario</font>

| Término | Significado |
| :-- | :-- |
| **Histograma** | Gráfico de barras pegadas que muestra la distribución de una variable cuantitativa. |
| **Distribución** | Forma en que se reparten los valores de una variable. |
| **Asimetría** | Falta de simetría; a la derecha si la cola larga va hacia valores altos. |
| **Bimodal** | Distribución con dos picos. |
