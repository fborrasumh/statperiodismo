## <font color='steelblue'>1. Media y mediana</font>

````python
salarios = pd.read_csv("salarios_redaccion.csv")
````

- La **media** ($\bar{x} = \frac{\sum x_i}{n}$) reparte el total a partes iguales: es lo que cobraría cada persona si se juntaran todos los sueldos y se dividieran.
- La **mediana** es el valor central cuando se ordenan los datos: la mitad cobra menos y la otra mitad más.

````python
s = salarios["salario_bruto_anual"]
print("Media:  ", round(s.mean()))
print("Mediana:", round(s.median()))
print("Personas por debajo de la media (%):", round((s < s.mean()).mean() * 100, 1))
````

**Qué ha pasado**: la media supera a la mediana en unos 2.500 euros, y más del 60 % de la plantilla cobra **menos que la media**. Los sueldos de dirección tiran de la media hacia arriba. Con distribuciones asimétricas a la derecha, **la mediana describe mejor al trabajador típico**.

````python
plt.hist(s, bins=25, color="lightgrey", edgecolor="white")
plt.axvline(s.mean(), color="firebrick", label=f"Media {s.mean():,.0f} €")
plt.axvline(s.median(), color="steelblue", label=f"Mediana {s.median():,.0f} €")
plt.legend()
plt.title("La media se va hacia la cola (datos simulados)")
plt.show()
````

## <font color='steelblue'>2. La moda y las variables cualitativas</font>

La **moda** es el valor más frecuente. Es la única medida de centralización que sirve para variables **nominales** (no hay «medio principal medio»).

````python
encuesta = pd.read_csv("encuesta_medios.csv")
print("Moda del medio principal:", encuesta["medio_principal"].mode()[0])
print("Moda de la confianza:", encuesta["confianza_medios"].mode()[0])
````

¿Qué medida usar?

| Tipo de variable | Medida |
| :-- | :-- |
| Nominal | Moda |
| Ordinal | Mediana (y moda) |
| Cuantitativa simétrica | Media (≈ mediana) |
| Cuantitativa asimétrica o con valores extremos | Mediana |

## <font color='steelblue'>3. Media ponderada</font>

Cuando los datos tienen **distinto peso**, la media simple engaña. Ejemplo: el share medio de un programa a lo largo de la semana no es la media de los shares diarios si cada día lo ve un número de personas distinto.

$$\bar{x}_w = \frac{\sum w_i x_i}{\sum w_i}$$

Las encuestas también usan pesos para que la muestra represente a la población (lo verás en el cuaderno 25).

````python
share_dia = np.array([12.0, 11.5, 13.2, 12.8, 14.1, 9.0, 8.5])
espectadores_tv = np.array([9.5, 9.3, 9.6, 9.4, 8.8, 7.2, 7.5])   # millones viendo la tele a esa hora
print("Media simple:   ", round(share_dia.mean(), 2))
print("Media ponderada:", round(np.average(share_dia, weights=espectadores_tv), 2))
````

**Qué ha pasado**: el fin de semana el share es más bajo pero hay menos gente delante del televisor; ponderar le da menos peso y la media sube un poco.

## <font color='steelblue'>4. Comparar grupos</font>

**Qué vamos a hacer**: comparar media y mediana del salario por tipo de contrato y por sexo. `agg()` calcula varias medidas a la vez.

````python
salarios.groupby("contrato")["salario_bruto_anual"].agg(["count", "mean", "median"]).round(0)
````

````python
por_sexo = salarios.groupby("sexo")["salario_bruto_anual"].median()
brecha = (por_sexo["Hombre"] - por_sexo["Mujer"]) / por_sexo["Hombre"] * 100
print(por_sexo)
print(f"Brecha salarial (sobre la mediana masculina): {brecha:.1f} %")
````

**Cuidado al publicarlo**: esta brecha compara a todas las mujeres con todos los hombres, sin tener en cuenta el puesto ni la experiencia. Es un dato real de la plantilla (simulada), pero no demuestra por sí solo que se pague distinto **por el mismo trabajo**. Lo correcto es decir qué se compara.

## <font color='steelblue'>Resumen: ideas clave</font>

- La **media** reparte el total por igual; la **mediana** deja la mitad de los casos a cada lado; la **moda** es el valor más frecuente.
- Con distribuciones **asimétricas** o con valores extremos, la mediana describe mejor el caso típico.
- Para variables **nominales** solo tiene sentido la moda.
- Si los datos tienen distinto peso, usa la **media ponderada** (`np.average(x, weights=w)`).
- Al comparar grupos, di exactamente **qué** se compara (todas las mujeres frente a todos los hombres, o a igual puesto).

## <font color='steelblue'>Glosario</font>

| Término | Significado |
| :-- | :-- |
| **Media** | Suma de los valores dividida por su número. |
| **Mediana** | Valor que deja la mitad de los datos por debajo. |
| **Moda** | Valor más frecuente. |
| **Media ponderada** | Media en la que cada valor cuenta según su peso. |
| **Brecha salarial** | Diferencia relativa entre los salarios de dos grupos. |
