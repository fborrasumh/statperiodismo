## <font color='steelblue'>1. Índices de cada producto</font>

````python
cesta = pd.read_csv("cesta_compra.csv")
cesta
````

La cesta (simulada) tiene precios y cantidades mensuales de un hogar en 2020 (base) y 2024. Empezamos por el índice simple de precios de cada producto:

````python
cesta["indice_precio"] = (cesta["precio_2024"] / cesta["precio_2020"] * 100).round(1)
cesta[["producto", "precio_2020", "precio_2024", "indice_precio"]].sort_values("indice_precio")
````

**Cómo leerlo**: el aceite de oliva tiene un índice de 228,6 (su precio se ha multiplicado por 2,3). El pan ha subido un 15 %. La media simple de los índices daría a todos los productos el mismo peso, como si gastáramos lo mismo en café que en pollo.

````python
print("Media simple de los índices:", round(cesta["indice_precio"].mean(), 1))
````

## <font color='steelblue'>2. Índice de Laspeyres</font>

El **índice de Laspeyres** compara lo que costaría **la cesta del año base** a los precios actuales con lo que costaba en el año base:

$$L = \frac{\sum p_t \, q_0}{\sum p_0 \, q_0} \times 100$$

Responde a: «¿cuánto más cuesta hoy comprar lo mismo que comprábamos en 2020?».

````python
gasto_base = (cesta["precio_2020"] * cesta["cantidad_2020"]).sum()
misma_cesta_hoy = (cesta["precio_2024"] * cesta["cantidad_2020"]).sum()
laspeyres = misma_cesta_hoy / gasto_base * 100
print(f"Gasto en 2020: {gasto_base:.2f} €   Misma cesta en 2024: {misma_cesta_hoy:.2f} €")
print(f"Índice de Laspeyres: {laspeyres:.1f}")
````

Es equivalente a una **media ponderada** de los índices simples, con pesos iguales al gasto de cada producto en el año base:

````python
pesos = cesta["precio_2020"] * cesta["cantidad_2020"]
print(round(np.average(cesta["indice_precio"], weights=pesos), 1))
````

## <font color='steelblue'>3. Índice de Paasche</font>

El **índice de Paasche** usa la cesta **actual**:

$$P = \frac{\sum p_t \, q_t}{\sum p_0 \, q_t} \times 100$$

Si los hogares han reaccionado comprando menos de lo que más ha subido (menos aceite, menos periódicos de papel), Paasche dará una subida **menor** que Laspeyres.

````python
paasche = (cesta["precio_2024"] * cesta["cantidad_2024"]).sum() / (cesta["precio_2020"] * cesta["cantidad_2024"]).sum() * 100
fisher = np.sqrt(laspeyres * paasche)
print(f"Laspeyres: {laspeyres:.1f}   Paasche: {paasche:.1f}   Fisher (media geométrica): {fisher:.1f}")
````

**Por qué el IPC usa Laspeyres (encadenado)**: solo necesita conocer las cantidades del periodo base, que se obtienen de una encuesta de gasto cara y lenta; los precios sí se recogen cada mes. Para que la cesta no se quede vieja, el INE actualiza las ponderaciones cada año (**Laspeyres encadenado**).

## <font color='steelblue'>Resumen: ideas clave</font>

- Un **índice compuesto** combina varios precios ponderando cada uno por su importancia.
- **Laspeyres** usa las cantidades del **año base**: ¿cuánto cuesta hoy la cesta de antes?
- **Paasche** usa las cantidades **actuales**; con efecto sustitución da subidas menores.
- Laspeyres equivale a una media de índices simples ponderada por el **gasto en el año base**.
- El **IPC** es un Laspeyres encadenado con ponderaciones que se actualizan cada año.

## <font color='steelblue'>Glosario</font>

| Término | Significado |
| :-- | :-- |
| **Índice compuesto** | Índice que resume la evolución de varios precios o cantidades. |
| **Laspeyres** | Índice ponderado con las cantidades del periodo base. |
| **Paasche** | Índice ponderado con las cantidades del periodo actual. |
| **Efecto sustitución** | Cambio de las compras hacia productos que suben menos. |
