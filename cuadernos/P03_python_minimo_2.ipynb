## <font color='steelblue'>1. Bucles: repetir sin copiar y pegar</font>

Un bucle `for` recorre los elementos de una lista y ejecuta el bloque sangrado una vez para cada uno.

````python
cadenas = ["Cadena 1", "Cadena 2", "Cadena 3"]
for cadena in cadenas:
    print("Analizando", cadena)
````

**Qué vamos a hacer**: calcular la tasa de variación mensual del paro para varios meses seguidos. `range(1, n)` genera los números 1, 2, ..., n−1.

````python
parados = [182450, 179630, 176200, 171900, 168050]
meses = ["enero", "febrero", "marzo", "abril", "mayo"]

for i in range(1, len(parados)):
    tasa = (parados[i] - parados[i - 1]) / parados[i - 1] * 100
    print(f"{meses[i]}: {tasa:.1f} %")
````

**Qué ha pasado**: en cada vuelta `i` vale 1, 2, 3 y 4, y comparamos cada mes con el anterior (`i - 1`). Enero no tiene tasa porque no hay mes previo.

## <font color='steelblue'>2. Funciones: cálculos que reutilizas</font>

Una **función** guarda un cálculo con un nombre. Se define con `def`, recibe **argumentos** entre paréntesis y devuelve un resultado con `return`.

````python
def tasa_variacion(inicial, final):
    """Devuelve la tasa de variación en %, con un decimal."""
    return round((final - inicial) / inicial * 100, 1)

print(tasa_variacion(12500, 15000))
print(tasa_variacion(182450, 179630))
````

Una función puede usar a otra. Esta escribe el titular completo:

````python
def titular_paro(lugar, inicial, final):
    t = tasa_variacion(inicial, final)
    verbo = "sube" if t > 0 else "baja"
    return f"El paro {verbo} un {abs(t)} % en {lugar}"

print(titular_paro("Alicante", 182450, 179630))
print(titular_paro("Castellón", 41200, 42150))
````

**Qué ha pasado**: `verbo = "sube" if t > 0 else "baja"` es una forma corta del `if` en una sola línea. Con dos funciones ya hemos automatizado un titular.

## <font color='steelblue'>3. Diccionarios: datos con etiqueta</font>

Un **diccionario** guarda pares **clave → valor** entre llaves. Es ideal para datos con nombre, como la población de cada ciudad.

````python
poblacion = {"Alicante": 349282, "Elche": 235580, "Torrevieja": 89000}
print(poblacion["Elche"])
poblacion["Orihuela"] = 80000          # añadir una clave nueva
for ciudad, habitantes in poblacion.items():
    print(f"{ciudad}: {habitantes} habitantes")
````

## <font color='steelblue'>4. Librerías: no reinventes la rueda</font>

Una **librería** es un conjunto de funciones ya escritas. Se cargan con `import`. Las que usaremos en todo el curso son:

| Librería | Para qué | Cómo se importa |
| :-- | :-- | :-- |
| `statistics` | Medidas sencillas (media, mediana) | `import statistics` |
| NumPy | Cálculo con muchos números | `import numpy as np` |
| Pandas | Tablas de datos | `import pandas as pd` |
| matplotlib | Gráficos | `import matplotlib.pyplot as plt` |

````python
import statistics
sueldos = [21000, 23500, 24800, 26100, 27000, 31000, 95000]
print("Media:", round(statistics.mean(sueldos)))
print("Mediana:", statistics.median(sueldos))
````

**Qué ha pasado**: un solo sueldo muy alto (95.000) arrastra la media hasta 35.486, mientras que la mediana (26.100) describe mejor al trabajador típico. Lo estudiaremos en el cuaderno 8.

````python
import pandas as pd
import matplotlib.pyplot as plt

serie = pd.Series(sueldos, name="sueldo")
serie.plot(kind="bar", color="steelblue")
plt.title("Siete sueldos inventados")
plt.show()
````

## <font color='steelblue'>Resumen: ideas clave</font>

- Un bucle **`for`** repite un bloque para cada elemento; `range(a, b)` va de a hasta b−1.
- Una **función** (`def nombre(argumentos): ... return resultado`) guarda un cálculo para reutilizarlo.
- Un **diccionario** guarda pares clave → valor y se consulta con `d[clave]`.
- Las **librerías** se cargan con `import`; en el curso usaremos `pd`, `np` y `plt`.
- Con valores extremos, la **mediana** describe mejor el caso típico que la media.

## <font color='steelblue'>Glosario</font>

| Término | Significado |
| :-- | :-- |
| **Bucle** | Instrucción que repite un bloque de código. |
| **Función** | Bloque de código con nombre que recibe datos y devuelve un resultado. |
| **Argumento** | Dato que se le pasa a una función. |
| **Diccionario** | Colección de pares clave → valor. |
| **Librería** | Conjunto de funciones listas para usar. |
