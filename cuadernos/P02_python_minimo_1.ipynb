## <font color='steelblue'>1. Variables y operaciones</font>

Una **variable** es un nombre que guarda un valor. Se crea con `=`. Elige nombres que expliquen qué guardan, sin tildes ni espacios (`parados_enero`, no `p1`).

````python
parados_enero = 182450
parados_febrero = 179630

diferencia = parados_febrero - parados_enero
print("Diferencia absoluta:", diferencia)
````

Las operaciones básicas son `+`, `-`, `*`, `/` y `**` (potencia). Con ellas calculamos la **tasa de variación**, la cifra más usada en prensa:

$$\text{tasa de variación (\%)} = \frac{\text{valor final} - \text{valor inicial}}{\text{valor inicial}} \times 100$$

````python
tasa = (parados_febrero - parados_enero) / parados_enero * 100
print("Tasa de variación:", tasa, "%")
print("Redondeada:", round(tasa, 1), "%")
````

**Qué ha pasado**: el paro ha bajado un 1,5 %. `round(valor, 1)` deja un decimal: en prensa casi nunca se publican más de uno o dos.

⚠️ Python usa **punto** como separador decimal (`1.5`), no coma.

## <font color='steelblue'>2. Textos y titulares automáticos</font>

Los textos (tipo `str`) van entre comillas. Una **f-string** (una `f` delante de las comillas) permite meter variables dentro del texto con llaves `{}`. Así se escriben los titulares automáticos de muchas redacciones.

````python
provincia = "Alicante"
tasa = -1.53
titular = f"El paro baja un {abs(tasa):.1f} % en {provincia}"
print(titular)
````

**Qué ha pasado**: `{abs(tasa):.1f}` escribe el valor absoluto de la tasa con un decimal. `abs()` quita el signo porque el verbo «baja» ya lo indica.

Algunas operaciones útiles con textos:

````python
medio = "  prensa Digital "
print(medio.strip())          # quita espacios al principio y al final
print(medio.strip().lower())  # todo en minúsculas
print(medio.strip().title())  # Mayúscula Inicial
print(len("Estadística"))     # número de caracteres
````

Limpiar textos así es habitual: en los datos reales «Prensa digital», «prensa digital » y «PRENSA DIGITAL» serían tres categorías distintas si no los unificamos.

## <font color='steelblue'>3. Listas: varios datos juntos</font>

Una **lista** guarda varios valores en orden, entre corchetes. Es la forma más sencilla de guardar, por ejemplo, la audiencia de una semana.

````python
audiencia = [1520, 1610, 1480, 1730, 1390, 980, 1100]   # miles de espectadores, de lunes a domingo
print("Días:", len(audiencia))
print("Total semanal:", sum(audiencia))
print("Máximo:", max(audiencia), "Mínimo:", min(audiencia))
print("Media diaria:", round(sum(audiencia) / len(audiencia), 1))
````

Cada elemento tiene una **posición** que empieza en **0**. `audiencia[0]` es el lunes y `audiencia[-1]` el último (domingo). Con `[a:b]` se toma un trozo, desde `a` hasta `b` **sin incluir** `b`.

````python
print("Lunes:", audiencia[0])
print("Domingo:", audiencia[-1])
print("Fin de semana:", audiencia[5:7])
````

Para añadir un dato usamos `append()`:

````python
audiencia.append(1650)   # el lunes siguiente
print(audiencia)
````

## <font color='steelblue'>4. Condiciones: decidir con los datos</font>

Con `if` el código hace una cosa u otra según se cumpla una condición. Las comparaciones son `>`, `<`, `>=`, `<=`, `==` (igual) y `!=` (distinto). Las líneas que dependen del `if` van **sangradas** (con 4 espacios).

````python
tasa = -1.53

if tasa > 0:
    verbo = "sube"
elif tasa < 0:
    verbo = "baja"
else:
    verbo = "se mantiene"

print(f"El paro {verbo} un {abs(tasa):.1f} %")
````

**Qué ha pasado**: como la tasa es negativa, se cumple la segunda condición y el verbo es «baja». Prueba a cambiar `tasa` por `2.4` o por `0` y vuelve a ejecutar.

Las condiciones se combinan con `and` (las dos) y `or` (al menos una):

````python
share = 14.2
franja = "Prime time"
if share > 12 and franja == "Prime time":
    print("Programa líder en su franja")
````

## <font color='steelblue'>Resumen: ideas clave</font>

- Una **variable** guarda un valor: `nombre = valor`. Usa nombres descriptivos.
- La **tasa de variación** es (final − inicial) / inicial × 100. Redondea con `round()`.
- Las **f-strings** (`f"... {variable:.1f} ..."`) meten cifras en un texto con el formato que quieras.
- Una **lista** guarda datos en orden; las posiciones empiezan en **0** y en `[a:b]` el final no se incluye.
- `if` / `elif` / `else` ejecutan un bloque u otro según una condición; el bloque va sangrado.

## <font color='steelblue'>Glosario</font>

| Término | Significado |
| :-- | :-- |
| **Variable** | Nombre que guarda un valor. |
| **`str`** | Tipo de dato de texto. |
| **f-string** | Texto con `f` delante donde lo que va entre llaves se sustituye por su valor. |
| **Lista** | Colección ordenada de valores entre corchetes. |
| **Tasa de variación** | Cambio relativo entre dos valores, en porcentaje. |

## <font color='steelblue'>Errores frecuentes y cómo resolverlos</font>

| Síntoma | Causa probable | Solución |
| :-- | :-- | :-- |

| `IndentationError` | El bloque del `if` no está sangrado | Pon 4 espacios delante de las líneas del bloque |
| `IndexError: list index out of range` | Pides una posición que no existe | Recuerda que la última posición es `len(lista) - 1` |
